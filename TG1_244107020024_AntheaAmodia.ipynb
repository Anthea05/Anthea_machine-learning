1. pyPREP merupakan library Python yang digunakan khusus untuk pra-pemrosesan (preprocessing) data sinyal otak (EEG). Fungsinya untuk membersihkan data mentah EEG sebelum dianalisis lebih lanjut, misalnya mendeteksi channel yang rusak atau noise, interpolasi channel yang bermasalah, dan referencing ulang sinyal. PyPREP ini adalah port dari pipeline standar bernama PREP.
